```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
df = df[['survived','sex', 'age', 'fare','embarked', 'class']]
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df_cat_enc = pd.get_dummies(df[['sex','embarked','class']],dtype='int',drop_first=True)
```

```python
final_df = pd.concat([df[['survived','age', 'fare']],df_cat_enc],axis=1)
```

```python
X = final_df.drop(columns='survived')
y = final_df['survived']
xtrain,xtest,ytrain,ytest = train_test_split(X,y,test_size=0.3,random_state=100)
```

```python
# scaling 
# build the model 
from sklearn.preprocessing import StandardScaler
SS = StandardScaler()
scaled_xtrain = SS.fit_transform(xtrain)
scaled_xtest = SS.transform(xtest)
```

```python
LR = LogisticRegression(max_iter=1000)
LR.fit(scaled_xtrain,ytrain)
lr_pred = LR.predict(scaled_xtest)
print(classification_report(ytest,lr_pred))
```
